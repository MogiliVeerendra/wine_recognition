```python
import pandas as pd
import numpy as np
import joblib as jb
```

```python
file_path = "../../Data/raw/wine.data"
column_names = [
    "Class",
    "Alcohol",
    "Malic_Acid",
    "Ash",
    "Alcalinity_of_Ash",
    "Magnesium",
    "Total_Phenols",
    "Flavanoids",
    "Nonflavanoid_Phenols",
    "Proanthocyanins",
    "Color_Intensity",
    "Hue",
    "OD280_OD315",
    "Proline"
]
wine = pd.read_csv(file_path,names = column_names, header = None)
wine
```

```python
print(wine[wine["Alcohol"] < 0].sum())

Q1 = wine["Alcohol"].quantile(0.25)
Q3 = wine["Alcohol"].quantile(0.75)

IQR = Q3 - Q1
lower_bound = Q1 - 1.5 * IQR
higher_bound = Q3 + 1.5 * IQR

wine[(wine["Alcohol"] < lower_bound) | (wine["Alcohol"] > higher_bound)]
```
